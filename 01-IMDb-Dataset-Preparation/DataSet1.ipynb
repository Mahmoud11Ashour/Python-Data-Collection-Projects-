```python
# Step 1: Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# This version uses relative file paths for local execution.
```

```python
# Step 2: Data Collection
# Load the IMDB dataset
data = pd.read_csv('IMDB-Movie-Data.csv')
```

```python
# Display initial rows of the dataset
data.head()
```

```python
# Step 3: Data Exploration
# Display basic information about the dataset
data.info()
```

```python
# Summary statistics for numerical features
data.describe()
```

```python
# Display column names and types for feature descriptions

print("Columns in the dataset:")
print(data.columns)
```

```python
# Step 4: Data Cleaning

# Check for missing values
print("Missing values in the dataset:")
print(data.isnull().sum())
```

```python
# Handle missing values with median for numerical columns
data['Revenue (Millions)'].fillna(data['Revenue (Millions)'].median(), inplace=True)
data['Metascore'].fillna(data['Metascore'].median(), inplace=True)
print(data.isnull().sum())
```

```python
# Check for duplicated values
print("duplicated values in the dataset:")
print(data.duplicated().sum())
```

```python
# Handle duplicated values
data.drop_duplicates(inplace=True)
#data.drop_duplicates()

print(data.duplicated().sum())
```

```python
# Step 5: Data Transformation

from sklearn.preprocessing import MinMaxScaler, StandardScaler, LabelEncoder
#*********************************************************************
# Normalize or scale numeric data using MinMaxScaler
#scaler = MinMaxScaler()
#data[['Rating', 'Revenue (Millions)', 'Metascore']] = scaler.fit_transform(data[['Rating', 'Revenue (Millions)', 'Metascore']])
#*********************************************************************
#  (Standardization) 'Rating', 'Revenue'
scaler = StandardScaler()
numerical_columns = ['Rating', 'Revenue (Millions)', 'Votes']
data[numerical_columns] = scaler.fit_transform(data[numerical_columns])
#*********************************************************************
data['Genre'] = data['Genre'].astype(str)
data['Genre'] = data['Genre'].str.split(',')

genre_dummies = data['Genre'].explode().str.get_dummies()

genre_dummies = genre_dummies.groupby(genre_dummies.index).sum()

data = pd.concat([data, genre_dummies], axis=1)

#*********************************************************************
encoder = LabelEncoder()
data['Director'] = encoder.fit_transform(data['Director'])
#*********************************************************************
#data['Genre'] = LabelEncoder().fit_transform(data['Genre'])
#*********************************************************************
# Encode categorical variables (Genre)
#data = pd.get_dummies(data, columns=['Genre'], drop_first=True) #One-Hot Encoding
```

```python
# Display column names and types for feature descriptions

print("Columns in the dataset:")
print(data.columns)
```

```python
data.info()
print('='*100)
print(data.isnull().sum())
print('='*100)
```

```python
# Step 6: Feature Engineering

# Example: Extract year from 'Title' if applicable or create a feature such as 'Actor_Count'
data['Actor_Count'] = data['Actors'].apply(lambda x: len(x.split(", ")))

# Create new feature: "Movie duration in hours"
data['Runtime_in_Hours'] = data['Runtime (Minutes)'] / 60
```
